```python
import numpy as  np
import matplotlib.pyplot as plt

from quilt.builder import SpikingNetwork, ParametricSpikingNetwork
from quilt.interface.base import set_verbosity

set_verbosity(0)
sn = ParametricSpikingNetwork.from_yaml("spiking_data/thalamus_loop_data/network.yaml", "spiking_data/thalamus_loop_data/neurons.yaml",
                                        ["spiking_data/thalamus_loop_data/test_strength.yaml"])
sn.monitor_spikes()
sn.monitor_states()
sn.set_parameters(strength=-0.5)

for proj in sn.features['projections'].values():
    print(proj['weight'], end=" ", flush=True)
```

```python
sn.run(dt=0.1, time=1000)
```

```python
frs = sn.get_firing_rates(delta_ms=1, sliding=False)

plt.plot(frs["TRN"])
```

```python
states = sn.get_neuron_states()
```

```python
for i in range(4):
    plt.plot(states['TRN'][i, :, 0], color='k')
```

```python
for i in range(4):
    plt.plot(states['M.RS'][i, :, 1], color='k')
```
